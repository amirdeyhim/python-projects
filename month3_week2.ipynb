```python
import pandas as pd
import numpy as np  

data = {
    "User": ["Ali", "Reza", "Sara", "Neda"],
    "Age": [25, np.nan, 30, 22],
    "City": ["Tehran", "Shiraz", np.nan, "Tehran"],
}
df = pd.DataFrame(data)

df
```

```python
clean_dropped = df.dropna()

print("--- خروجی Dropna ---")
print(clean_dropped)
print("\n" + "=" * 30 + "\n")

clean_filled = df.copy()  
clean_filled["Age"] = clean_filled["Age"].fillna(0)
clean_filled["Age"] = clean_filled["Age"].astype(int)
clean_filled["City"] = clean_filled["City"].fillna("Unknown")

print("--- خروجی Fillna ---")
print(clean_filled)
```

```python
import pandas as pd

data = {
    "User_ID": ["U1", "U1", "U2", "U3"],
    "Name": ["Ali", "Ali", "Reza", "Sara"],
    "Amount": [500, 500, 1200, 300],
}
df_dup = pd.DataFrame(data)

print("--- قبل از حذف تکراری‌ها ---")
print(df_dup)

clean_dup = df_dup.copy()
clean_dup = clean_dup.drop_duplicates()

print("\n--- بعد از حذف تکراری‌ها ---")
print(clean_dup)
```

```python
import pandas as pd

data = {
    "User_ID": ["U1", "U1", "U2"],
    "Amount": [500, 500, 1200],
    "Time": ["10:01", "10:02", "10:05"],  # کلیک دوم یک دقیقه بعد ثبت شده
}
df_real = pd.DataFrame(data)

print("--- تلاش برای حذف بدون Subset (اشتباه) ---")
print(df_real.drop_duplicates())

print("\n--- تلاش برای حذف با Subset (درست) ---")
clean_real = df_real.drop_duplicates(subset=["User_ID", "Amount"])
print(clean_real)
```

```python
import pandas as pd

data = {
    "Order_ID": [101, 102, 103, 104, 105],
    "User_ID": ["U1", "U1", "U2", "U1", "U3"],
    "Product": ["Laptop", "Laptop", "Mouse", "Laptop", "Keyboard"],
    "Time": ["10:00:01", "10:00:03", "10:15:00", "15:30:00", "16:00:00"],
}
df_orders = pd.DataFrame(data)
clean_orders = df_orders.copy()
clean_orders = df_orders.drop_duplicates(subset=["User_ID", "Product"], keep="last")
clean_orders
```

```python
import pandas as pd

data = {"Order_ID": [1, 2, 3], "Price": ["1,000$", "2,500$", "500$"]}
df_price = pd.DataFrame(data)

print("--- دیتای خام ---")
print(df_price)

clean_price = df_price.copy()

clean_price["Price"] = clean_price["Price"].str.replace("$", "")

clean_price["Price"] = clean_price["Price"].str.replace(",", "")

clean_price["Price"] = clean_price["Price"].astype(int)

print("\n--- دیتای تمیز شده برای محاسبات ریاضی ---")
print(clean_price)
```

```python
import pandas as pd

data = {
    "Campaign_ID": ["C1", "C2", "C3"],
    "Revenue": ["1,500,000 T", "850,000 T", "2,300,000 T"],
}
df_campaign = pd.DataFrame(data)
Clean_campaign = df_campaign.copy()
Clean_campaign["Revenue"] = Clean_campaign["Revenue"].str.replace(",", "").str.replace(" T", "")
Clean_campaign["Revenue"] = Clean_campaign["Revenue"].astype(int)
Clean_campaign
```

```python
import pandas as pd

data = {
    "User_ID": ["U1", "U2", "U3"],
    "Join_Date": ["2023-10-01", "05/12/2023", "2023.08.15"],
}
df_dates = pd.DataFrame(data)

print("--- قبل از تبدیل (همه فقط متن هستند) ---")
print(df_dates)

clean_dates = df_dates.copy()

clean_dates["Join_Date"] = pd.to_datetime(clean_dates["Join_Date"], format="mixed")
print("\n--- بعد از استانداردسازی تاریخ ---")
print(clean_dates)
```

```python
import pandas as pd

data = {
    "User_ID": ["U1", "U2", "U3"],
    "Join_Date": ["2023-10-01", "2023/10/05", "2023-10-10"],
    "First_Purchase_Date": ["2023-10-05", "2023/10/20", "2023-10-11"],
}
df_users = pd.DataFrame(data)

clean_users = df_users.copy()
clean_users["Join_Date"] = pd.to_datetime(clean_users["Join_Date"], format="mixed")
clean_users["First_Purchase_Date"] = pd.to_datetime(clean_users["First_Purchase_Date"], format="mixed")
clean_users
clean_users["Days_to_Purchase"] = (
    clean_users["First_Purchase_Date"] - clean_users["Join_Date"]
)
clean_users
```

```python
import pandas as pd
import numpy as np

data = {
    "Order_ID": [101, 102, 102, 103, 104],
    "User_ID": ["U1", "U2", "U2", "U3", "U4"],
    "Date": ["2023-05-01", "2023/05/05", "2023/05/05", "2023.05.10", "2023-05-15"],
    "Revenue": ["1,000$", "2,500$", "2,500$", np.nan, "500$"],
}
df_final = pd.DataFrame(data)
clean_final = df_final.copy()
clean_final = clean_final.drop_duplicates(subset=["Order_ID", "User_ID"])
clean_final["Revenue"] = clean_final["Revenue"].fillna("0")
clean_final["Revenue"] =(clean_final["Revenue"].str.replace("$", "").str.replace(",", "").astype(int))
clean_final["Date"]=pd.to_datetime(clean_final["Date"], format="mixed")
clean_final
```
